```python
# 혹시 파일이 아니라 폴더가 계속 이어지는 구조인지 전체 트리 얕게 확인
def print_tree(path, max_depth=6, prefix=""):
    if max_depth == 0:
        return
    try:
        entries = sorted(os.listdir(path))
    except NotADirectoryError:
        return
    for entry in entries[:5]:  # 각 레벨에서 5개까지만
        full_path = os.path.join(path, entry)
        print(prefix + entry + ("/" if os.path.isdir(full_path) else ""))
        if os.path.isdir(full_path):
            print_tree(full_path, max_depth - 1, prefix + "  ")

print_tree("/content/h2o_pose")
```

```python
import numpy as np

def parse_h2o_hand_pose(txt_path, n_joints=21):
    """
    반환: {
        'left': (21,3) array 또는 None(존재 안 함),
        'right': (21,3) array 또는 None
    }
    """
    with open(txt_path, 'r') as f:
        values = list(map(float, f.read().strip().split()))

    assert len(values) == 128, f"예상치 못한 길이: {len(values)}"

    left_flag = values[0]
    left_coords = np.array(values[1:64]).reshape(n_joints, 3)

    right_flag = values[64]
    right_coords = np.array(values[65:128]).reshape(n_joints, 3)

    return {
        'left': left_coords if left_flag > 0.5 else None,
        'right': right_coords if right_flag > 0.5 else None,
    }

# 테스트
sample_path = "/content/h2o_pose/subject2/k1/0/cam4/hand_pose/000000.txt"
result = parse_h2o_hand_pose(sample_path)
print("왼손 존재:", result['left'] is not None)
print("오른손 존재:", result['right'] is not None)
if result['right'] is not None:
    print("오른손 shape:", result['right'].shape)
    print("첫 관절(손목?) 좌표:", result['right'][0])
```

## 1. 모든 take 순회해서 시퀀스 수집

```python
import glob, os
import numpy as np

def find_all_takes(root="/content/h2o_pose"):
    """hand_pose 폴더가 있는 모든 경로 찾기"""
    return glob.glob(os.path.join(root, "*/*/*/cam4/hand_pose"))

def load_take_sequence(hand_pose_dir, n_joints=21):
    """
    한 take 폴더 안의 모든 프레임을 시간순으로 읽어서
    {'left': (frames,21,3) or None, 'right': (frames,21,3) or None} 반환
    (연속성이 깨지면 None 반환 — 아까 확인했듯 대부분 연속이지만 방어적으로)
    """
    files = sorted(os.listdir(hand_pose_dir), key=lambda x: int(x.replace(".txt", "")))
    frame_numbers = [int(f.replace(".txt", "")) for f in files]

    if frame_numbers != list(range(frame_numbers[0], frame_numbers[-1] + 1)):
        return None  # 연속 아니면 스킵 (이번엔 안전장치로만)

    left_seq, right_seq = [], []
    left_valid, right_valid = True, True

    for f in files:
        path = os.path.join(hand_pose_dir, f)
        parsed = parse_h2o_hand_pose(path, n_joints)
        if parsed['left'] is not None:
            left_seq.append(parsed['left'])
        else:
            left_valid = False
        if parsed['right'] is not None:
            right_seq.append(parsed['right'])
        else:
            right_valid = False

    result = {}
    # 전체 시퀀스 내내 손이 계속 존재해야 깨끗한 연속 시퀀스로 취급
    result['left'] = np.stack(left_seq) if (left_valid and len(left_seq) == len(files)) else None
    result['right'] = np.stack(right_seq) if (right_valid and len(right_seq) == len(files)) else None
    return result
```

```python
take_dirs = find_all_takes()
print(f"전체 take 수: {len(take_dirs)}")

all_h2o_sequences = []
skipped = 0

for take_dir in take_dirs:
    result = load_take_sequence(take_dir)
    if result is None:
        skipped += 1
        continue
    if result['left'] is not None:
        all_h2o_sequences.append(result['left'])
    if result['right'] is not None:
        all_h2o_sequences.append(result['right'])

print(f"추출된 손 시퀀스 수: {len(all_h2o_sequences)}")
print(f"스킵된 take 수: {skipped}")

lengths = [s.shape[0] for s in all_h2o_sequences]
print(f"최소: {min(lengths)}, 최대: {max(lengths)}, 평균: {np.mean(lengths):.1f}")
```

## 2. 정규화 (관절 순서 불확실성 감안한 안전한 방식)

```python
def normalize_h2o_sequence(seq, wrist_idx=0):
    """
    seq: (frames, 21, 3), MANO 컨벤션(0=손목) 가정
    스케일: 손목-전체관절 평균 거리로 정규화 (특정 손가락 인덱스에 안 의존)
    """
    wrist = seq[:, wrist_idx:wrist_idx+1, :]
    relative = seq - wrist

    dists = np.linalg.norm(relative, axis=-1)  # (frames, 21)
    scale = dists.mean(axis=-1, keepdims=True)  # (frames, 1) - 손목 자신(거리 0) 포함해도 무방
    scale = np.clip(scale, 1e-6, None)

    normalized = relative / scale[:, :, None]
    return normalized
```

## 3. Train/Val 분리 (subject 기준)

```python
def get_subject_from_path(take_dir):
    # /content/h2o_pose/subject2/k1/0/cam4/hand_pose -> "subject2"
    parts = take_dir.split(os.sep)
    return [p for p in parts if p.startswith("subject")][0]

train_sequences, val_sequences = [], []

for take_dir in take_dirs:
    subject = get_subject_from_path(take_dir)
    result = load_take_sequence(take_dir)
    if result is None:
        continue
    target_list = val_sequences if subject == "subject4" else train_sequences
    for hand in ["left", "right"]:
        if result[hand] is not None:
            target_list.append(result[hand])

print(f"train: {len(train_sequences)}개, val: {len(val_sequences)}개")
```

## 4. 마스킹 (기존 함수 재사용)

시퀀스 하나하나가 매우 길어서 시퀀스에서 여러 구간을 뽑기로 함.

```python
L_values = [5, 10, 20, 30]

def create_masked_sample(seq, L, N=15, rng=None):
    rng = rng or np.random.default_rng()
    total_needed = N + L
    if seq.shape[0] < total_needed:
        return None
    start = rng.integers(0, seq.shape[0] - total_needed + 1)
    input_seq = seq[start : start+N]
    target_seq = seq[start+N : start+N+L]
    return input_seq, target_seq

def build_pretrain_pairs_nonoverlap(sequences, L_values, N=15):
    """
    각 시퀀스를 (N+L) 크기로 겹치지 않게 순서대로 잘라서
    최대한 많은 쌍을 뽑음 (랜덤 샘플링 없음, 결정적)
    """
    final = {}

    for L in L_values:
        total_needed = N + L
        pairs = []

        for seq in sequences:
            if seq.shape[0] < total_needed:
                continue

            norm_seq = normalize_h2o_sequence(seq)
            n_windows = norm_seq.shape[0] // total_needed  # 겹치지 않게 몇 개 들어가는지

            for i in range(n_windows):
                start = i * total_needed
                input_seq = norm_seq[start : start + N]
                target_seq = norm_seq[start + N : start + N + L]
                pairs.append((input_seq, target_seq))

        final[L] = pairs
        avg_per_seq = len(pairs) / len(sequences)
        print(f"L={L}: {len(pairs)}개 쌍 생성 (시퀀스당 평균 {avg_per_seq:.1f}개)")

    return final

print("=== Train (non-overlapping) ===")
h2o_train_pairs = build_pretrain_pairs_nonoverlap(train_sequences, L_values)
print("=== Val (non-overlapping) ===")
h2o_val_pairs = build_pretrain_pairs_nonoverlap(val_sequences, L_values)
```
